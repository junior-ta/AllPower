```python
from pathlib import Path
from IPython.display import display, Image
from allpower.vision import classify

for label in ("normal", "defect"):
    photo = sorted(
        Path(f"data/vision_demo/test/{label}").glob("*.png")
    )[0]

    print(f"\nACTUAL LABEL: {label.upper()}")
    display(Image(filename=str(photo), width=240))
    print("MODEL RESULT:", classify(photo.read_bytes()))
```

```python
import csv
from pathlib import Path

photo = sorted(Path("data/vision_demo/test/defect").glob("*.png"))[0]
relative_path = photo.relative_to("data/vision_demo").as_posix()

with open("data/vision_demo/manifest.csv", newline="") as file:
    record = next(
        row for row in csv.DictReader(file)
        if row["path"] == relative_path
    )

print(record["source"])
```
